## Análise do primeiro dataset (2025)

```python
import pandas as pd
```

```python
caminho = "../data/raw/CelularesSubtraidos_2025.xlsx"

with pd.ExcelFile(caminho) as arquivo:
    print(arquivo.sheet_names)
```

```python
df = pd.read_excel(caminho, sheet_name = "CELULAR_2025")
display(df.head())
```

```python
print("Formato do DataFrame (linhas, colunas): ")
display(df.shape)
print("\nInformações gerais: ")
df.info()
print("\nDescrição: ")
df.describe()
```

```python
# Valores faltantes e duplicados
print("Valores nulos por coluna:")
display(df.isna().sum())

print("\nRegistros duplicados: ")
display(df.duplicated().sum())
```

```python
df.columns
```

```python
df.dtypes
```

```python
df["NOME_DEPARTAMENTO"].value_counts()
```

```python
# Valores da variável Alvo - Target - Label
df["RUBRICA"].value_counts()
```

```python
total_furtos = (df["RUBRICA"] == "Furto (art. 155)").sum()
total_roubos = (df["RUBRICA"] == "Roubo (art. 157)").sum()

print(f"Número de furtos: {total_furtos}")
print(f"Número de roubos: {total_roubos}")
```
